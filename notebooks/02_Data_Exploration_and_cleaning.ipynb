```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
```

```python
apple = yf.download(
    "AAPL",
    period="5y",
    multi_level_index=False
)
apple.head()
```

```python
apple["Close"].plot()
plt.title("Apple Stock Price (Last 5 Years)")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.show()
```

```python
import os
print(os.getcwd())
apple.to_csv("../data/raw/apple_stock.csv")
```
