```python
import pandas as pd
import matplotlib.pyplot as plt

apple = pd.read_csv("../data/processed/apple_stock_cleaned.csv")
nvidia = pd.read_csv("../data/processed/nvidia_stock_cleaned.csv")
microsoft = pd.read_csv("../data/processed/microsoft_stock_cleaned.csv")

apple["Date"] = pd.to_datetime(apple["Date"])
nvidia["Date"] = pd.to_datetime(nvidia["Date"])
microsoft["Date"] = pd.to_datetime(microsoft["Date"])
```

```python
plt.figure(figsize=(12,6))
plt.plot(apple["Date"], apple["Close"], label="Apple")
plt.plot(nvidia["Date"], nvidia["Close"], label="NVIDIA")
plt.plot(microsoft["Date"], microsoft["Close"], label="Microsoft")
plt.xlabel("Date")
plt.ylabel("Closing Price (USD)")
plt.title("Apple vs NVIDIA vs Microsoft Stock Prices")
plt.legend()
plt.grid()

plt.show()

# Observations
# All three companies show an overall upward trend over the 5 year period, even though their stock prices go up and down along the way.
# Microsoft has the highest stock price throughout most of the time period and shows steady long-term growth.
# Apple also shows consistent growth with fewer sudden changes, making its stock price appear more stable than nvidia's.
# Nvidia has the fastest increase in stock price compared to where it started. It also shows larger ups and downs, meaning its stock price is more volatile.
# All three companies experience short-term drops at different times, but each recovers and continues its overall upward trend.
# Overall, the graph shows that Apple, Microsoft, and Nvidia have all grown over time, but each company has a different growth pattern and level of volatility.
```

```python
apple["Daily Return"] = (apple["Close"] - apple["Close"].shift(1)) / apple["Close"].shift(1)

nvidia["Daily Return"] = (nvidia["Close"] - nvidia["Close"].shift(1)) / nvidia["Close"].shift(1)
microsoft["Daily Return"] = (microsoft["Close"] - microsoft["Close"].shift(1)) / microsoft["Close"].shift(1)

print(nvidia["Daily Return"].describe())
# nvidia’s average daily return was about 0.246 %.
# Its daily volatility was about 3.275 %.
# Its worst day was about -16.968 %.
# Its best day was about 24.369 %.

print(microsoft["Daily Return"].describe())
# microsoft’s average daily return was about 0.065 %.
# Its daily volatility was about 1.777 %.
# Its worst day was about -9.993 %.
# Its best day was about 15.507 %.

print(apple.columns)
print(nvidia.columns)
print(microsoft.columns)
```

```python
plt.figure(figsize=(12,6))

plt.plot(apple["Date"], apple["Daily Return"], label="Apple")
plt.plot(nvidia["Date"], nvidia["Daily Return"], label="NVIDIA")
plt.plot(microsoft["Date"], microsoft["Daily Return"], label="Microsoft")
plt.xlabel("Date")
plt.ylabel("Daily Return")
plt.title("Daily Returns Comparison")
plt.legend()
plt.grid()
plt.axhline(y=0, color="black", linestyle="--", linewidth=1)
plt.show()

# Observations
# All three stocks have daily returns that fluctuate around 0, meaning their prices increase and decrease from one trading day to the next.
# NVIDIA shows the largest positive and negative spikes, indicating that it experiences the highest daily volatility among the three companies.
# Apple has relatively smaller fluctuations, suggesting more stable daily price movements compared to NVIDIA.
# Microsoft also fluctuates around zero, but its daily returns are generally less extreme than NVIDIA's and similar to Apple's.
# Most daily returns for all three companies remain close to 0, while the larger spikes represent days with significant market movements or company-specific news.
```
