```python
# RSI (Relative Strength Index) is a technical indicator that measures the strength of recent price movements. 
# It ranges from 0 to 100 and helps identify whether a stock may be overbought or oversold.
# In simple words tells us how strong recent buying or selling has been.

# RSI scale 
# Above 70 (Overbought)- price may fall
# Under 30 (Oversold) - price may rise
# Around 50 - neutral 
# 

apple["Change"]=apple["Close"].diff() # daily price changes
apple["Gain"]=apple["Change"].where(apple["Change"] > 0, 0) # positive price changes
apple["Loss"]=-apple["Change"].where(apple["Change"] < 0, 0) # negative price chnages

avg_gain = apple["Gain"].rolling(window=14).mean()  # avg gains over 14 trading periods
avg_loss = apple["Loss"].rolling(window=14).mean() # avg losses over 14 trading periods

rs = avg_gain / avg_loss

apple["RSI"] = 100 - (100 / (1 + rs))
```
